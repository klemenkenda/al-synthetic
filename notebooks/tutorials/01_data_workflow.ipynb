# Data Workflow and Visualization

Focused on the `synth_surface_defects` dataset located under `data/`. We'll load metadata, display example images by label, and show basic statistics.

## 1. Setup Paths and Imports
Read CSV metadata and import necessary libraries.

```python
import pandas as pd
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

base = Path('../../data/synth_surface_defects')
meta_path = base / 'metadata.csv'
assert meta_path.exists(), f'{meta_path} not found'

meta = pd.read_csv(meta_path)
print('Loaded metadata', meta.shape)
meta.head()

# check for embeddings
emb_dir = Path('../../artifacts/embedding_cnn/embeddings')
if not emb_dir.exists():
    print('\nWARNING: embeddings directory not found at', emb_dir)
    print('You will need to train the embedding CNN first by running:')
    print('  python -m src.train.train_embedding_cnn --data-root data/synth_surface_defects')
    print('or use the provided Docker/commands from README to produce embeddings.')
else:
    print('\nEmbeddings are available; you can load them for modeling later.')
```

## 2. Dataset Statistics
Show counts per label and some basic information.

```python
# counts by label
display(meta['label'].value_counts())

# basic stats
display(meta.describe(include='all'))
```

## 3. Show Example Images by Label
Select one random image per label and display them.

```python
labels = meta['label'].unique()
fig, axes = plt.subplots(1, len(labels), figsize=(3*len(labels),3))
for ax, lab in zip(axes.flatten(), labels):
    row = meta[meta['label']==lab].sample(1, random_state=1).iloc[0]
    img = Image.open(base/row['filepath'])
    ax.imshow(img, cmap='gray')
    ax.set_title(lab)
    ax.axis('off')
plt.tight_layout()
```

## 4. Clustered or Grouped Views
Display several images from two different labels (clusters).

```python
def show_samples(lbl, n=6):
    subset = meta[meta['label']==lbl].sample(n, random_state=2)
    fig, axes = plt.subplots(1, n, figsize=(n*2,2))
    for ax, (_, row) in zip(axes, subset.iterrows()):
        img = Image.open(base/row['filepath'])
        ax.imshow(img, cmap='gray')
        ax.axis('off')
    fig.suptitle(lbl)
    plt.tight_layout()

show_samples('crack', 6)
show_samples('scratch',6)
```
