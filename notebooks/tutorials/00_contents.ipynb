# Tutorial Syllabus

This notebook lists the sequence of tutorial notebooks we will use in the workshop. Each notebook builds on the previous one and is aimed at students with minimal programming or AI background.

## Notebook Index

1. **01_data_workflow.ipynb** – Loading and exploring data using pandas and plotting
2. **02_simple_model.ipynb** – Training a tiny classifier, visualizing results
3. **03_active_learning.ipynb** *(optional)* – Introduction to active learning concepts and simulation


### How to Use This Syllabus

- Open each notebook (01–03) in your preferred environment (Jupyter, VS Code, Docker container, etc.).
- Work through them in sequence; exercises are embedded in the notebooks.
- All content is self-contained; no dependence on Google Drive or Colab.
- The earlier notebooks (setup, python basics, repo walkthrough) have been skipped for this run, so start with dataset exploration.
